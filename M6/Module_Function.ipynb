## Argument of function

```python
def adder(*args: int)-> None:
    for i in args:
        print(i);

adder(1,2,3,4,5)        
```

```python
def adder(a: int,b:int,c:int)-> None:
    print(f"{a} : {b} : {c}")   
```

Positional Arguments: Values are passed in order, e.g., adder(1, 2, 3).

```python
adder(1,2,3)
```

Keyword Arguments: Values are passed by parameter name, so order doesn't matter.

```python
adder(c=3,b=2,a=1)     
```

**kwargs (Arbitrary Keyword Arguments): 


```python
# Accepts multiple named arguments as a dictionary (key-value pairs).
# kwargs: int means all keyword argument values are expected to be integers.
def adder(**kwargs: int)-> None:
    for key,value in kwargs.items():
        print(f"{key} : {value}");

adder(a=1,b=2,c=3,n_estimator=50);        
```

# Unpacking

```python
def double_maney(firstPerson: int,secouondPersion:int) -> tuple[int,int]:
    return firstPerson*2, secouondPersion *2;

# unzip
akash,bikash = double_maney(10,20)
print(akash,bikash)
```

```python
# Returns multiple values as a tuple.
# *rest collects the remaining returned values into a list.
# This is called Extended Iterable Unpacking.

def double_money(a: int, b: int, c: int, d: int) -> tuple[int, ...]:
    return a * 2, b * 2, c * 2, d * 2

first, *rest = double_money(10, 20, 30, 40)

print(first)  # 20
print(rest)   # [40, 60, 80]
```

## Generator

```python
data_list=[x for x in range(500)] # list comprehension
def data_loader(chunk_size:int,list:list[int]):
    for i in range(0,len(list),chunk_size):
        yield list[i:i+chunk_size];

generator_data = data_loader(5,data_list);        
print(next(generator_data))
print(next(generator_data))
```
