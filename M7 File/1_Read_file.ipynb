# File Cursor Position and Index Using `tell()`

- `tell()` returns the current position of the file cursor.
- `read(n)` reads up to `n` characters from the current cursor position.
- `read()` reads all remaining content from the current cursor position.
- After reading, the file cursor moves forward automatically.

```python
with open("../symple.txt","r") as file:
    print(file.tell())  # Initial cursor position: 0
    print(file.read(5))  # Read the first 5 characters
    print(file.tell())  # Cursor position after reading 5 characters: 5
    print(file.read(), end="")  # Read all remaining content (default size: -1)
    print(file.tell())  # Final cursor position after reading the entire file
```

# Python File Handling: `seek()` and `tell()`

- `tell()` returns the current file cursor position.
- `seek(5)` moves the cursor to position 5.
- `read()` reads all remaining content from the current cursor position.
- `seek(0)` resets the cursor to the beginning of the file.
- After `seek(0)`, `read()` reads the entire file again.
- `end=""` prevents `print()` from adding an extra newline.

```python
with open("../symple.txt", "r") as file:
    print(file.tell())  # Initial cursor position: 0
    print(file.seek(5))  # Move the cursor to position 5
    print(file.tell())  # Current cursor position: 5
    print(file.read(), end="")  # Read all remaining content from position 5
    file.seek(0)  # Reset the cursor to the beginning of the file
    print(file.read(), end="")  # Read the entire file again from position 0
```
