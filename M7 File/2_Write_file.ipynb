### File Direct Write / Replace (`w` Mode)

Python-এ `w` mode ব্যবহার করলে ফাইলে আগে থেকে থাকা সব content মুছে যায় এবং নতুন করে লেখা content দিয়ে পুরো ফাইল replace হয়। ফাইলটি না থাকলে নতুন ফাইল তৈরি হয়।

```python
with open("./write_test.txt","w") as file:
    file.write("Hello World\n")
    file.write("My name is sampod\n")
    
with open("./write_test.txt","r") as file:
    content = file.read();
    print(content)
```

## apend context

```python
with open("./write_test.txt","a") as file:
    file.write("Hello World append\n");
    file.write("My name is sampod append\n");

with open("./write_test.txt","r") as file:
    content = file.read();
    print(content)    
```

```python
mystring = ["Hello World","My name is sampod"]
with open("./write_test2.txt","w") as file:
    file.writelines(mystring) 
```
