While Loop

```python
balance =10;
while balance > 0 :
    print("balance is", balance);
    balance = balance -1;
wife_salary = 1000;
while wife_salary > 0 :
    print("wife salary is", wife_salary);
    wife_salary = wife_salary -100;    
```
